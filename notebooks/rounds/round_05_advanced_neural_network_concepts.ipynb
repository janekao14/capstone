# Round 5: Advanced neural-network concepts

This notebook is the reproducible round record. It reads the canonical query history, limits the evidence to observations available at this stage and recalculates each function's weekly incumbent.

## Decision context

The Round 5 strategy used the course concept **Advanced neural-network concepts** as a diagnostic layer. The living documentation records the submitted vector, returned output, rationale and reflection.

```python
from pathlib import Path
import csv

ROOT = Path.cwd().resolve().parents[1] if Path.cwd().name == 'rounds' else Path.cwd().resolve()
HISTORY = ROOT / 'data' / 'query_history.csv'
with HISTORY.open() as stream:
    history = list(csv.DictReader(stream))
print(f'Loaded {len(history)} evaluated weekly queries from {HISTORY.relative_to(ROOT)}')
```

```python
ROUND = 5
available = [row for row in history if int(row['round']) <= ROUND]
print('function | best weekly output | first achieved | query')
for function in range(1, 9):
    rows = [row for row in available if int(row['function']) == function]
    best = max(rows, key=lambda row: float(row['output']))
    coordinates = [best[f'x{i}'] for i in range(1, 9) if best[f'x{i}']]
    print(
        f"{function:>8} | {float(best['output']):>18.12g} | "
        f"{int(best['round']):>14} | [{', '.join(coordinates)}]"
    )
```

## Interpretation

The table above is calculated only from information available through this round. Later observations are excluded, preventing hindsight from changing the historical incumbent.
